```python
#Changing none to np.nan

df.replace({'none': np.nan}, inplace=True)

df['management'].unique()
```

```python
#Dropping all nan

df.dropna(inplace=True)

df.info()
df.isnull().sum()
```

```python
#handling inconsistency in education type and management
df.replace({
    'integrated': 'Integrated',
    'formal_only': 'Formal',
    'religious_only': 'Religious',
    'dk': 'DK',
    'public': 'Public',
    'private': 'Private',
    'faith_based': 'Faith Based'
    

}, inplace=True)

df['education_type'].unique()
```

```python
#Converting object to a boolean

object = ['improved_water_supply', 'improved_sanitation', 'chalkboard', 'PHCN_electricity']
df[object] = (df[object].astype(bool))

df.info()
```

```python
#Handling date type

df['survey_date'] = pd.to_datetime(df['survey_date'], errors="coerce")

df.info()
```

```python
#Creating total teachers column
df['total_teachers'] = df['male_teachers'] + df['female_teachers']
df.head()
```

```python
#Creating total teachers column
# df['total_students'] = df['male_students'] + df['female_students']
# df.head()
```

```python
#Creating Student-to-Teacher Ratio Column

df['student_teacher_ratio'] = round(df['total_student'] / df['total_teachers'])
df.head()

```

```python
#Creating State Column

df['states'] = df['LGA'].str.split('_').str[0].str.title()
df.head()
```

```python
#Facility density by location

df['LGA_facility_density'] = df.groupby('LGA')['facility_name'].transform('count')
df.head()

```

```python
#Dealing with Outliers
numeric = ['NCE_teachers', 'full_time_teachers', 'total_student', 'total_classrooms', 'total_teachers', 'student_teacher_ratio', 'total_toilets', 'male_students', 'female_students', 'male_teachers', 'female_teachers']

#Detecting outliers using the ".quantile()"
Q1 = df[numeric].quantile(0.25)
Q3 = df[numeric].quantile(0.75)

IQR = Q3 - Q1

#Finding the Lower and upper bounds
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR

outliers = (df[numeric] < lower_bound) | (df[numeric] > upper_bound)

print('Outliers:', outliers.sum())


```

```python
# Creating a new dataset while excluding outliers

# numeric = ['NCE_teachers', 'full_time_teachers', 'total_student', 'total_classrooms', 'total_teachers', 'student_teacher_ratio', 'total_toilets', 'male_students', 'female_students', 'male_teachers', 'female_teachers']
# lower_bound = Q1 - 1.5*IQR
# upper_bound = Q3 + 1.5*IQR

# new_df = df[
#     (df[numeric] > lower_bound) & (df[numeric] < upper_bound)
# ]

# new_df


# Calculate Q1, Q3, and IQR for all numeric columns
Q1 = df.select_dtypes(include=["number"]).quantile(0.25)
Q3 = df.select_dtypes(include=["number"]).quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Keep rows where ALL numeric features fall within the bounds
numeric_cols = df.select_dtypes(include=["number"]).columns
condition = ((df[numeric_cols] >= lower_bound) & (df[numeric_cols] <= upper_bound)).all(axis=1)

new_df = df[condition]
new_df
```

Part 4: Data Visualization
Create clear and well-labeled visualizations using Plotly. At minimum, include:
    Bar chart of school types
    Histogram of student population
    Boxplot of teacher distribution
    Pie chart of water and sanitation access
    Map visualization using latitude and longitude


```python
#Bar chart of school types

school_bar = new_df.groupby('facility_type')['facility_name'].count().sort_values(ascending=False)
fig1 = px.bar(
    x=school_bar.values,
    y=school_bar.index,
    title='Type of Schools'

)
fig1.update_layout(xaxis_title='Count', yaxis_title='School Type')
fig1.show()
```

```python
#Histogram of student population

fig2 = px.histogram(
    new_df,
    x='total_student',
    nbins=20,
    title='Student Population'
)
fig2.update_traces(
    marker_line_color='red',
    marker_line_width=1
)
fig2.show()

```

```python
#Boxplot of teacher distribution

fig3 = px.box(
    new_df,
    y='total_teachers',
    title="Teachers' Distribution"
)

fig3.show()
```

```python
# Pie chart of water access
water_access = new_df['improved_water_supply'].value_counts()
fig4 = px.pie(
    values=water_access.values,
    names=water_access.index,
    title='Distribution of Water Access'
)
fig4.show()
```

```python
# Pie chart of sanitation access
sanitation_access = new_df['improved_sanitation'].value_counts()
fig4 = px.pie(
    values=sanitation_access.values,
    names=sanitation_access.index,
    title='Distribution of Sanitation Access'
)
fig4.show()
```

```python
#Map visualization using latitude and longitude

cord = ['latitude', 'longitude']
corr = new_df[cord].corr()

sns.heatmap(corr, annot=True, cmap="vlag", fmt='.2f')

plt.title('Heatmap of the Cordinates')
plt.show
```
